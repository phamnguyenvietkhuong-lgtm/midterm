````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ==========================================
# 第 2 格：建立資料庫與初始資料
# ==========================================
import sqlite3

# 建立或連接 SQLite 資料庫
conn = sqlite3.connect('activities.db')
cursor = conn.cursor()

# 每次執行都 DROP TABLE IF EXISTS 重建
cursor.execute('DROP TABLE IF EXISTS 報名紀錄')
cursor.execute('DROP TABLE IF EXISTS 活動')

# 建立活動表
cursor.execute('''
CREATE TABLE 活動 (
    活動名稱 TEXT PRIMARY KEY,
    日期 TEXT,
    總名額 INTEGER
)
''')

# 建立報名紀錄表
cursor.execute('''
CREATE TABLE 報名紀錄 (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    姓名 TEXT,
    活動名稱 TEXT,
    人數 INTEGER,
    狀態 TEXT
)
''')

# 放入初始資料
initial_activities = [
    ('迎新晚會', '2026-11-20', 6),
    ('程式工作坊', '2026-11-25', 4),
    ('籃球友誼賽', '2026-11-28', 3)
]

cursor.executemany('INSERT INTO 活動 (活動名稱, 日期, 總名額) VALUES (?, ?, ?)', initial_activities)
conn.commit()
conn.close()
print("✅ 資料庫與初始資料建立完成！")
````

````python
# ==========================================
# 第 4 格：AI 解讀函數
# ==========================================
import json
import time
import re

def parse_intent(user_input):
    prompt = f"""
你是一個活動系統助理。請分析學生的輸入句子，並嚴格以 JSON 格式回傳以下三個欄位：
1. "動作": 限填 "報名"、"取消"、"查詢" 或 "無法判斷"
2. "活動": 限填 "迎新晚會"、"程式工作坊"、"籃球友誼賽" 或 "無法判斷"
3. "人數": 數字 (包含同行友人，例如 "我和兩個朋友" = 3；若未提及則預設為 1，若超過範圍程式會自行檢查)

使用者輸入：「{user_input}」

請直接回傳 JSON 格式，不要包含其他多餘文字。格式範例:
{{"動作": "報名", "活動": "迎新晚會", "人數": 1}}
"""

    for attempt in range(3):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                reasoning_effort=REASONING_EFFORT
            )
            raw_text = response.choices[0].message.content

            match = re.search(r'\{.*\}', raw_text, re.DOTALL)
            if match:
                data = json.loads(match.group(0))
                action = data.get("動作", "無法判斷")
                activity = data.get("活動", "無法判斷")
                count = int(data.get("人數", 1))
                return action, activity, count, raw_text
        except Exception as e:
            if "429" in str(e):
                time.sleep(30)
                continue
            time.sleep(2)

    return "無法判斷", "無法判斷", 1, "API 呼叫失敗或逾時"
````

````python
# ==========================================
# 第 5 格：AI 解讀測試 (8~10句)
# ==========================================
test_sentences = [
    "我想報名迎新晚會",
    "幫我和兩個朋友報名程式工作坊",
    "我要取消籃球友誼賽",
    "請問迎新晚會還有名額嗎",
    "幫我報名 5 個人參加籃球賽",
    "我想要買章魚燒",
    "取消程式工作坊",
    "查一下籃球友誼賽",
    "幫我報名迎新晚會",
    "我要查迎新晚會"
]

results = []
for idx, text in enumerate(test_sentences, 1):
    action, activity, count, raw = parse_intent(text)
    results.append({
        "#": idx,
        "使用者說的話": text,
        "預期動作": action,
        "預期對象(活動)": activity,
        "預期數量": count,
        "AI原始回覆": raw
    })

df_test = pd.DataFrame(results)
display(df_test)
````

````python
# ==========================================
# 第 6 格：功能函數
# ==========================================
def process_user_request(name, user_input):
    if not name or not name.strip():
        return "⚠️ 請先輸入您的姓名！"

    action, activity, count, _ = parse_intent(user_input)

    if action == "無法判斷" or activity == "無法判斷":
        conn = sqlite3.connect('activities.db')
        cursor = conn.cursor()
        cursor.execute("SELECT 活動名稱, 日期, 總名額 FROM 活動")
        acts = cursor.fetchall()
        conn.close()
        act_list = "\n".join([f"- {a[0]} (日期: {a[1]}, 總名額: {a[2]})" for a in acts])
        return f"🤖 無法辨識您的意圖。目前的活動清單如下：\n{act_list}"

    conn = sqlite3.connect('activities.db')
    cursor = conn.cursor()
    try:
        # A1: 檢查人數是否在 1~3 之間
        if not (1 <= count <= 3):
            return f"❌ 報名失敗：人數 ({count}) 不符合規定，每次報名人數必須介於 1 到 3 人之間。"

        if action == "報名":
            cursor.execute("SELECT 總名額 FROM 活動 WHERE 活動名稱 = ?", (activity,))
            act_row = cursor.fetchone()
            if not act_row:
                return f"❌ 查無此活動：{activity}"
            total_slots = act_row[0]

            cursor.execute("SELECT id FROM 報名紀錄 WHERE 姓名 = ? AND 活動名稱 = ? AND 狀態 = '已報名'", (name, activity))
            if cursor.fetchone():
                return f"❌ 報名失敗：{name} 已經報名過「{activity}」了，不可重複報名。"

            # A2: 檢查個人報名上限（最多同時報名 2 個活動）
            cursor.execute("SELECT COUNT(*) FROM 報名紀錄 WHERE 姓名 = ? AND 狀態 = '已報名'", (name,))
            active_count = cursor.fetchone()[0]
            if active_count >= 2:
                return f"❌ 報名失敗：每位學生最多同時報名 2 個活動（您目前已有 {active_count} 個有效報名）。"

            cursor.execute("SELECT SUM(人數) FROM 報名紀錄 WHERE 活動名稱 = ? AND 狀態 = '已報名'", (activity,))
            sum_row = cursor.fetchone()[0]
            current_booked = sum_row if sum_row else 0

            remaining = total_slots - current_booked
            if count > remaining:
                return f"❌ 報名失敗：「{activity}」剩餘名額不足（剩餘 {remaining} 人，欲報名 {count} 人）。"

            cursor.execute("INSERT INTO 報名紀錄 (姓名, 活動名稱, 人數, 狀態) VALUES (?, ?, ?, '已報名')", (name, activity, count))
            conn.commit()
            return f"✅ 報名成功！{name} 成功報名「{activity}」，人數：{count} 人。"

        elif action == "取消":
            cursor.execute("SELECT id FROM 報名紀錄 WHERE 姓名 = ? AND 活動名稱 = ? AND 狀態 = '已報名'", (name, activity))
            reg_row = cursor.fetchone()
            if not reg_row:
                return f"❌ 取消失敗：找不到 {name} 報名「{activity}」的有效紀錄。"

            cursor.execute("UPDATE 報名紀錄 SET 狀態 = '已取消' WHERE id = ?", (reg_row[0],))
            conn.commit()
            return f"✅ 已成功取消「{activity}」的報名紀錄！"

        elif action == "查詢":
            cursor.execute("SELECT 總名額 FROM 活動 WHERE 活動名稱 = ?", (activity,))
            act_row = cursor.fetchone()
            if not act_row:
                return f"❌ 查無此活動：{activity}"
            total_slots = act_row[0]

            cursor.execute("SELECT SUM(人數) FROM 報名紀錄 WHERE 活動名稱 = ? AND 狀態 = '已報名'", (activity,))
            sum_row = cursor.fetchone()[0]
            current_booked = sum_row if sum_row else 0
            remaining = total_slots - current_booked
            return f"📊 查詢結果：「{activity}」總名額 {total_slots} 人，已報名 {current_booked} 人，剩餘名額：{remaining} 人。"

    except Exception as e:
        return f"⚠️ 系統錯誤：{str(e)}"
    finally:
        conn.close()

# A3: 統計查詢函數 (使用 JOIN 與 GROUP BY)
def get_statistics():
    conn = sqlite3.connect('activities.db')
    query = """
        SELECT
            a.活動名稱,
            a.日期,
            a.總名額 AS 名額,
            COALESCE(SUM(CASE WHEN r.狀態 = '已報名' THEN r.人數 ELSE 0 END), 0) AS 已報名人數,
            a.總名額 - COALESCE(SUM(CASE WHEN r.狀態 = '已報名' THEN r.人數 ELSE 0 END), 0) AS 剩餘名額
        FROM 活動 a
        LEFT JOIN 報名紀錄 r ON a.活動名稱 = r.活動名稱
        GROUP BY a.活動名稱, a.日期, a.總名額
    """
    df_stats = pd.read_sql_query(query, conn)
    conn.close()
    return df_stats

print("✅ 功能函數載入完成！")
````
